# 1. Filtering Rows
- Multiple condition filtering: we use (&,|,~): syntax-> df[(df['column']condition)(op) df(df['column']condition)]
- Single condition filtering: syntax -> df[ df['column'] condition ]

```python
import pandas as pd

# Sample DataFrame
df = pd.DataFrame({
    'name': ['Ali', 'Sara', 'Zain', 'Noor'],
    'age': [23, 30, 17, 45],
    'city': ['Lahore', 'Karachi', 'Islamabad', 'Lahore']
})
df
```

```python
df[(df['age']>20) & (df['city']!='Karachi')]
```

# 2.Creating columns
We can:
- Combine columns using math
- Create condition-based flags (like True/False)
- Use `.apply()` or NumPy for logic-based columns

```python
import pandas as pd

df = pd.DataFrame({
    'employee': ['Ali', 'Sara', 'Zain'],
    'hours_worked': [40, 35, 50],
    'hourly_rate': [500, 600, 550]
})
df['is_overtime'] = np.where(df['hours_worked'] > 40, True, False)
df['salary'] = df['hours_worked'] * df['hourly_rate']
df
```

```python
#Creating new column bonus 
df['bonus'] = np.where(df['is_overtime']==True,2000,0)
df
```

```python
df['weekly_tax'] = df['salary']*0.10
df
```

```python
df['net_pay'] = df['salary'] + df['bonus'] - df['weekly_tax']
df
```

# 3.handling_missing
### you'll learn how to:
- Detect missing values
- Drop missing da using dropna: syntax-> df.dropna(substet[column])
- Fill missing values (imputati using fillna: syntax -> df.fillna(value))
- Practice with real examples

```python
df = pd.DataFrame({
    'name': ['Ali', 'Sara', 'Zain', np.nan],
    'age': [25, np.nan, 30, 22],
    'city': ['Lahore', 'Karachi', np.nan, 'Lahore']
})
df
```

```python
# to count how many missing value we have in each column we use: .isnull().sum()
df.isnull().sum()
```

```python
#Drop rows where city has NaN value 
df = df.dropna(subset=['city'])
```

```python
#use fillna to fill missing value with no_name and age with it's median value 
df = df.fillna({
    'name':'no_name',
    'age':df['age'].median()
})
```

```python
#final check to make sure if there is not missing values:
print(df.isnull().sum())
print(df)
```

# 4. encoding
Machine learning models don’t understand text — they need numbers.

We use **encoding** to convert categorical values into numeric values.

We’ll cover:
- Label Encoding (use .map to assign either 1 or 0 to a categorical column) use only when dealing with columns like gender
- One-Hot Encoding (use get_dummies)
- When to use each


```python
df = pd.DataFrame({
    'name': ['Ali', 'Sara', 'Zain', 'Noor'],
    'gender': ['Male', 'Female', 'Male', 'Female'],
    'department': ['HR', 'Finance', 'IT', 'HR']
})
df
```

```python
#label encoding 
df['gender'] = df['gender'].map({'Male':0,'Female':1})
```

```python
#one-hot encoding 
df = pd.get_dummies(df,columns=['department'], dtype=int)
```

```python
df
```

# 5. Binning Continuous Variables

Binning helps convert continuous values into categorical buckets.

Why?
- Makes patterns easier to spot
- Reduces noise
- Sometimes improves model performance
- Use Pd.cut or np.where

```python
df = pd.DataFrame({
    'name': ['Ali', 'Sara', 'Zain', 'Noor', 'Hassan', 'Ayesha'],
    'age': [17, 22, 36, 42, 29, 64]
})
df
```

```python
# using np.where 
df['age_group_where'] = np.where(df['age']<20,'Teen',
                           np.where(df['age']<40,'Adult','senior'))
```

```python
#using pd.cut 
bins = [0,20,40,100]
labels = ['Teen','Adult','Senior']
df['age_group_cut'] = pd.cut(df['age'],right=False,bins=bins, labels=labels)
```

### What is Feature Scaling?

Feature scaling means **bringing numeric values (like age, salary, height)** into the **same range** so that no one feature dominates others during model training.

It’s super important when using models like:
- KNN
- SVM
- Logistic Regression
- Neural Networks

---

### Two Common Techniques

| Method            | What it does                            | Range       |
|------------------|------------------------------------------|-------------|
| 🔹 Min-Max Scaling | Scales values between 0 and 1            | 0 → 1       |
| 🔺 Z-score Scaling | Centers values around mean 0, std 1      | ~ -3 → +3   |

You’ll learn more about these when we start **machine learning models** later.

## 7. Text Feature Engineering

Text columns often need cleaning before analysis or modeling.

We'll cover:
- Lowercasing and trimming
- String search
- Extracting parts
- Feature creation from text


```python
data = pd.DataFrame({
    'name': [' Ali Khan', 'SARA', 'Zain.', 'Noor Sheikh'],
    'email': ['ali@gmail.com', 'sara@outlook.com', 'zain@yahoo.com', 'noor@gmail.com'],
    'comment': ['Great!', 'BAD', 'Average', 'Excellent!!']
})
data
```

```python
data['name'] = data['name'].str.strip().str.lower() #Remove spaces(strip), convert text to lowercase (lower)
```

```python
data['Is_Gmail'] = data['email'].str.contains("@gmail.com")
```

```python
data['domain'] = data['email'].str.extract(r'@(.+)$')
```

```python
data['comment_sentiment'] = np.where(data['comment'].str.lower().str.contains('excellent|great'),'Positive',
                                    np.where(data['comment'].str.lower().str.contains('bad'),'Negative','Neutral'))
```

```python
data['comment'] = data['comment'].str.replace('!', '', regex=False)
```

```python
data
```

## String Cleaning Cheatsheet

- Remove extra spaces → `.str.strip()`
- Lowercase → `.str.lower()`
- Replace → `.str.replace('old', 'new')`
- Check for text → `.str.contains('keyword')`
- Extract → `.str.extract('regex')`
- Split → `.str.split('@')`


## 8. Datetime Feature Engineering

Time-based data is very common in real-world datasets.

We'll cover:
- Converting to datetime
- Extracting year, month, day
- Calculating durations
- Categorizing based on time


```python
import pandas as pd

df = pd.DataFrame({
    'employee': ['Ali', 'Sara', 'Zain', 'Noor', 'Ahmed'],
    'joining_date': ['2020-01-10', '2023-03-25', '2021-07-19', '2022-11-05', '2019-09-01']
})

df['joining_date'] = pd.to_datetime(df['joining_date'])  # Convert to datetime
df
```

```python
#Extract joining year,month,day, weekday 
df['Joining_year'] = df['joining_date'].dt.year
df['Joining_month'] = df['joining_date'].dt.month
df['Joining_day'] = df['joining_date'].dt.day
df['Joining_week'] = df['joining_date'].dt.day_name()
```

```python
#how many days since a person joined 
today = pd.Timestamp.today()
df['tenure_years'] = (today - df['joining_date']).dt.days // 365
```

```python
#add labels
bins = [0, 1, 2, float('inf')]
labels = ['New', 'Mid', 'Experienced']
df['tenure_label'] = pd.cut(df['tenure_years'], bins=bins, labels=labels)
```
