```python
import matplotlib.pyplot as plt 
import seaborn as sns
import numpy as np
import random
```

```python
df = sns.load_dataset("iris")
df
```

```python
fig, ax = plt.subplots(figsize=(10,4))
ax.plot(df['sepal_length'],color='red',linewidth=2,marker='+')
ax.set_xlabel('sample Index')
ax.set_ylabel('sepal length')
plt.title("Sepal Length Across Iris Samples", fontsize=14, fontweight='bold')
ax.grid(True,linestyle=':',linewidth=0.5)
plt.tight_layout()
```

```python
#Multiple plots  
plt.figure(figsize=(10,3))
x = np.array([1, 2, 3, 4])
y = np.array([5, 6, 7, 8])

x1 = [1, 4, 6, 8]
y1 = [10, 12, 14, 16]

plt.plot(x, y)
plt.plot(x1, y1)

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Multiple Plots demo using lineplot', fontsize=12, fontweight='bold')
plt.grid(True)
plt.show()
```

```python
#Barplot representation
Fruits= ['Apple', 'Banana', 'Orange', 'Mango', 'Grapes']
Sales = [150, 200, 120, 300, 180]
plt.figure(figsize=(10,6))
plt.bar(Fruits,Sales,width=0.7,color='lightgreen',edgecolor='black')
plt.xlabel("Fruits")
plt.ylabel("Sales")
plt.title("Fruits Sales",fontsize=12,fontweight='bold')
```

```python
#Multiple barplots
Product= ['Laptop', 'Tablet', 'Smartphone', 'Headphones']
Q1= [320, 180, 410, 260]
Q2= [300, 200, 430, 280]
Q3= [310, 210, 450, 300]
Q4= [330, 220, 470, 310]
barWidth = 0.5 
x = np.arange(len(Product))

plt.figure(figsize=(10,6))
plt.bar(x-1.5*barWidth,Q1,color='r',edgecolor='black',label='Company 1',width=0.3)
plt.bar(x-0.5*barWidth,Q2,color='blue',edgecolor='black',label='Company 2',width=0.3)
plt.bar(x+1.5*barWidth,Q3,color='navy',edgecolor='black',label='Company 3',width=0.3)
plt.bar(x+0.5*barWidth,Q4,color='green',edgecolor='black',label='Company 4',width=0.3)
plt.xticks(x,Product)
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Quarterly Sales by Product",fontweight='bold',fontsize=12)
plt.legend()
plt.show()
```

```python
#Scatter plot
n1 = np.array([3, 12, 9, 20, 5, 18, 22, 11, 27, 16])
n2 = np.array([95, 55, 63, 77, 89, 50, 41, 70, 58, 83])
a = [20, 50, 100, 200, 500, 1000, 60, 90, 150, 300] 
b = ['red', 'green', 'blue', 'purple', 'orange', 'black', 'pink', 'brown', 'yellow', 'cyan'] 
plt.scatter(n1,n2,c=b,s=a,marker='.',edgecolors='black',linewidths=1,alpha=0.6) 
plt.title("A Scatter Plot with Different Colors and their Size",fontsize=12,fontweight='bold')
#Here alpha is used to control transparency
# (s) is used to set marker size, (c) is used to give colors 
```

```python
#Histogram 
np.random.seed(42)
scores = np.random.normal(loc=70, scale=10, size=100).astype(int)
scores = np.clip(scores, 0, 100)
scores[:10]
plt.hist(scores,bins=10,color='skyblue',rwidth=0.8,edgecolor='black')
#Histplots in matplotlib are simple, to get more functionality on histplots use seaborn a library built over matplotlib that provides extra functionality
```

```python
#Making subplots -> it provides you functionality to build multiple graphs on one figure here (fig) is the figure on which we'll draw the graphs 
#and (axs) wil be the plots and (2,2) in the subplots are defining a 2D array where we plotted in the next line 
fig, axs = plt.subplots(2, 2)

axs[0, 0].plot([1, 2], [3, 4]) #[0,0] top left
axs[0, 1].bar([1, 2], [5, 6]) # [0,1] top right
axs[1, 0].hist([1, 1, 2, 3, 4]) # [1,0] bottom left 
axs[1, 1].pie([20, 80], labels=['Fail', 'Pass']) # [1,1] bottom right 
# works similar to indexing in arrays 
```

```python
import matplotlib.pyplot as plt

labels = ['Python', 'Java', 'C++', 'JavaScript']
sizes = [35, 25, 20, 20]
colors = ['#66b3ff','#99ff99','#ffcc99','#ff9999']
explode = [0.1, 0, 0, 0] 

plt.figure(figsize=(6, 6))
plt.pie(sizes,
        labels=labels,
        autopct='%.1f%%',
        startangle=140,
        shadow=True,
        explode=explode,
        colors=colors,
        wedgeprops={'edgecolor': 'black'},
        textprops={'fontsize': 12, 'color': 'black'})

plt.title('Programming Language Popularity',fontsize=12, fontweight='bold')
plt.show()
```

```python
#Grid in matplotlib -> Gives clarity while reading plots 
x1 = [1.5,2.5,3.5,5,7]
y1 = [1.5,2.5,3.5,5,7]

fig,(ax1,ax2) = plt.subplots(1,2, figsize=(12,5))

#Plot without a grid 
ax1.plot(x1,y1)
ax1.set_title("Plot without a Grid")

ax2.plot(x1,y1)
ax2.grid(True)
ax2.set_title("Plot with a Grid")
```
