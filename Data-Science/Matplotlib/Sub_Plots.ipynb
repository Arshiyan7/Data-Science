## Create 1 figure with 3 subplots side-by-side (1 row × 3 columns) that show:
- Histogram of sepal length
- Histogram of petal length
- Scatter plot of sepal length vs petal length

````python
#Load Dataset:
df = sns.load_dataset('iris')
````

````python
df.head()
````

````python
#create subplot 
fig, (ax1,ax2,ax3) = plt.subplots(nrows=1,ncols=3,figsize=(15, 4))
#Histogram of sepal length
ax1.hist(x=df['sepal_length'],bins=30,color='r',rwidth=0.9)
ax1.set_title('Histogram for Sepal length')
#Histogram of petal length 
ax2.hist(x=df['petal_length'],bins=30,color='navy',rwidth=0.9)
ax2.set_title('Histogram for petal length')
#Scatter plot of sepal length vs petal length
ax3.scatter(x=df['sepal_length'],y=df['petal_length'],color='crimson')
ax3.set_title("Sepal Length vs Petal Length")
plt.tight_layout()
plt.show()
````

````python
#Save Fig 
fig.savefig("Practice Plots")
````

# Understanding Subplots in Matplotlib

In data visualization, a single figure can contain multiple plots using subplots. This is useful when comparing different aspects of a dataset side-by-side.

The function `plt.subplots()` is commonly used to create subplots. It returns two objects: `fig` and `ax`. Here, `fig` represents the entire figure or canvas, and `ax` represents the individual plot or subplot inside the figure. When creating multiple subplots, you can unpack them like `fig, (ax1, ax2) = plt.subplots(nrows=2, ncols=1)`. In this example, `nrows=2` means there will be two rows of plots, and `ncols=1` means one column, so ax1 and ax2 will be stacked vertically.

The parameters `sharex` and `sharey` allow subplots to share axes. Setting `sharex=True` means the x-axis of all subplots will be shared, making their scales consistent. Similarly, `sharey=True` will share the y-axis among subplots. For example, `fig, (ax1, ax2) = plt.subplots(nrows=2, sharex=True)` will share the x-axis scale between ax1 and ax2.
