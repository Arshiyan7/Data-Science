```python
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data = sns.load_dataset("tips")
data
```

# Relational plots:

```python
#Line plot
plt.grid(True,linestyle='-',linewidth=0.5)
sns.lineplot(data=data, x='total_bill', y='tip',hue='sex', color='r',errorbar=None)
plt.title("line Plot of Total Bill vs tip", fontsize=14)
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.grid(True)
plt.show()
#Use errorbar to remove the fill between, try removing the error bar argument you may notice the background of line being filled
```

```python
data
```

```python
#2 Scatterplots
tips = sns.load_dataset("tips")
sns.set(style="whitegrid")

plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex", style="smoker", palette="Set2")

plt.title("Scatter Plot of Total Bill vs Tip", fontsize=14)
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.grid(True)
plt.show()
```

```python
#3 relplot
sns.set_style('whitegrid')
sns.relplot(data=data, x='total_bill',y='tip',hue='sex',style='smoker',col='sex',row='time')
#Give row and col to make multiple plots , and give title, x and y label when plotting only 1 plot 
```

# Categorical plot 

## Barplot

```python
#1 barplot  
import numpy as np
sns.set_style("darkgrid")
plt.title('total bill over days')
plt.xlabel('total bill')
plt.ylabel('Day')
sns.barplot(data=data, x='total_bill',y='day',palette='magma',estimator=np.mean,hue='day',legend=False)
```

## Boxplot

```python
sns.set_style("darkgrid")
plt.title('total bill over days')
plt.xlabel('total bill')
plt.ylabel('Day')
sns.boxplot(data=tips, x='day', y='total_bill', palette='magma', hue='day', legend=False)
```

## Violin plot 

```python
sns.set_style("darkgrid")
plt.title('total bill over days')
plt.xlabel('total bill')
plt.ylabel('Day')
sns.violinplot(data=tips, x='day', y='total_bill', hue='sex',split=True,inner='quartile')
```

# Distribution plots

## Histplot

```python
sns.set_style("darkgrid")
plt.title('total bill demo using Histplot',fontweight='bold',fontsize=12)
plt.xlabel('total bill')
sns.histplot(data=tips, x='total_bill',bins=30,color='skyblue')
```

## KDE plot

```python
sns.set_style("darkgrid")
plt.title('total bill demo using KDE plot',fontweight='bold',fontsize=12)
plt.xlabel('total bill')
sns.kdeplot(data=tips, x='total_bill',fill=True,hue='day')
```

## Pairplot 

```python
sns.set_style("darkgrid")
plt.title('total bill demo using Pairplot plot',fontweight='bold',fontsize=12)
plt.xlabel('total bill')
sns.pairplot(data=data, hue='sex')
```

```python
sns.jointplot(data=tips, x='total_bill', y='tip', kind='kde', fill=True)
```

# Regression plots 

## lmplot

```python
sns.lmplot(data=data,x='total_bill',y='tip',hue='sex')
```

```python
sns.lmplot(data=data,x='total_bill',y='tip',hue='sex',row='smoker',col='day',markers =['o', 'v'])
```

# Heatmap

```python
tips_corr = tips.corr(numeric_only=True)
sns.heatmap(data=tips_corr,cmap='coolwarm',annot=True)
```

```python
#Heatmap using pivot table  
pivot_tips = tips.pivot_table(values='total_bill',aggfunc='mean',columns='sex',index='day')
sns.heatmap(pivot_tips,annot=True,cmap='YlGnBu')
```
