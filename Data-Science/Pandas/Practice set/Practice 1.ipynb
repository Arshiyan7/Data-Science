```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'Score': [88, 92, 95, 70]
})
df
```

```python
import pandas as pd

data = {
    'Team': ['A', 'A', 'B', 'B', 'C', 'C'],
    'Player': ['P1', 'P2', 'P3', 'P4', 'P5', 'P6'],
    'Points': [10, 15, 20, 25, 30, 35],
    'Fouls': [1, 2, 0, 3, 2, 4]
}
df = pd.DataFrame(data)
df
```

```python
team = df.groupby('Team')['Points'].sum().reset_index()
fouls = df.groupby('Team')['Fouls'].mean().reset_index()
points = df.groupby('Team')['Points'].max().reset_index()
number = df.groupby('Team').size().reset_index(name='Player_Count')
agg = df.groupby('Team').agg({'Points': 'mean', 'Fouls': 'sum'}).reset_index()
agg
```

```python
import pandas as pd

data = {
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Product': ['A', 'A', 'B', 'C', 'B', 'B', 'A', 'A', 'C', 'C'],
    'Sales': [120, 150, 200, 130, 180, 160, 140, 170, 190, 110],
    'Discount': [5, 10, 7, 6, 5, 8, 9, 4, 6, 7],
    'Returns': [1, 0, 2, 1, 1, 3, 0, 1, 2, 2],
    'Cost': [80, 100, 120, 90, 110, 105, 85, 95, 115, 70]
}
df = pd.DataFrame(data)
df
```

```python
df['Profit'] = df['Sales'] - df['Cost']
profit_region = df.groupby('Region')['Profit'].sum().reset_index()
df['Return_rate'] = df['Returns'] / df['Sales']
avg_discount = df.groupby('Product').agg({'Discount':'mean','Return_rate':'mean'}).reset_index()
avg_profit = df.groupby(['Region','Product'])['Profit'].mean().reset_index()
avg_profit.sort_values(by='Region',ascending=True)
region_product = df.groupby(['Region','Product']).agg({'Sales':'sum','Cost':'mean'}).reset_index()
high = df.groupby(['Product','Region'])['Sales'].sum().reset_index()
high.max()
```
