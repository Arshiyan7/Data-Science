```python
import pandas as pd
import numpy as np

data = {
    'member_id': [101, 102, 103, 104, 105],
    'name': ['John Doe', 'Alice Smith', 'Bob Johnson', 'Eva Williams', 'Mike Brown'],
    'membership_type': ['Basic', 'Premium', 'Basic', 'Student', 'Premium'],
    'months_active': [3, 12, 7, 2, 24],
    'monthly_fee': [30, 50, 30, 20, 50]
}

df = pd.DataFrame(data)
print("Original Data:")
df
```

```python
df['total_paid'] = df['monthly_fee'] * df['months_active']
```

```python
df['loyalty'] = df['months_active'].apply(lambda x: 'Bronze' if x<6 else ('Silver' if x<12 else 'Gold'))
```

```python
discounts = {
    'Premium': 0.9,
    'Student': 0.8,
    'Basic': 1.0
}
df['disscounted_fees'] = df['monthly_fee'] * df['membership_type'].map(discounts)
```

```python
df[['first_name', 'last_name']] = df['name'].str.split(expand=True)
df = df.drop(columns=['name'])
```

```python
import pandas as pd
import numpy as np

data = {
    'product_id': ['A-100', 'B-200', 'C-300', 'D-400', 'E-500'],
    'product_name': ['Smartphone X', '  Wireless Earbuds', 'Smart Watch ', 'Tablet Pro  ', 'Laptop Ultra'],
    'category': ['Phones', 'Audio', 'Wearables', 'Tablets', 'Computers'],
    'current_price': [799.99, ' 129.99 ', 199.50, '349.00', '999.00'],
    'units_in_stock': [45, 102, 28, '15', 7],
    'last_restock_date': ['2023-05-15', '2023-06-01', 'Invalid', '2023-06-10', 'NaN']
}

df = pd.DataFrame(data)
print("Original Messy Data:")
df
```

```python
df['product_id'] = df['product_id'].str.strip()
df['current_price'] = df['current_price'].str.strip().astype(float)
df['units_in_stock'] = df['units_in_stock'].astype(int)
df['last_restock_date'] = df['last_restock_date'].replace(['Invalid','NaN',''],np.nan)
df['last_restock_date'] = pd.to_datetime(df['last_restock_date'])
```

```python
labels = ['Low','Medium','High']
bins = [0,20,50,np.inf]
df['stock_status'] = pd.cut(df['units_in_stock'],bins=bins,labels=labels)
lab = ['Budget','midrange','Premium']
bi = [0,200,800,np.inf]
df['price_tier'] = pd.cut(df['current_price'],bins=bi,labels=lab)
df['last_restock_date'] = df['last_restock_date'].replace(['Invalid','NaN',''],'Unknown')
```

```python
df.loc[df['price_tier'] == 'Premium', 'current_price'] *= 1.15
df['estimated_value'] = df['current_price'] * df['units_in_stock']
```

```python
import pandas as pd
import numpy as np

data = {
    'order_id': ['#1001', '1002', '#1003', '1004', '1005', '1006'],
    'customer': ['john@email.com', 'ALICE@GMAIL.COM', 'bob_123@yahoo.com', np.nan, 'invalid', 'DAVID@shop.com'],
    'product': ['Laptop|16GB', 'Mouse (Wireless)', 'Keyboard', 'Monitor-24"', 'USB-C Cable', 'Phone_Case'],
    'category': ['Electronics', 'Electronics', 'Electronics', 'Electronics', 'Accessories', None],
    'price': ['$999.99', '$25.50', '45.00', 200.00, '10.99', ''],
    'date': ['2023-01-15', 'Invalid', '2023-01-20', '2023-01-18', 'NaN', '2023-01-22'],
    'rating': [5, 4, None, 3, 2, 5]
}
df = pd.DataFrame(data)
df
```

```python
#Data Cleaning 
#df['order_id'] = df['order_id'].str.replace('#','').astype(int)
df['domains'] = df['customer'].str.extract(r'@([\w.-]+)$')
df['product'] = df['product'].str.replace(r'[^a-zA-Z0-9]','',regex=True)
'''df['price'] = (df['price']
               .str.replace('$','',regex=False)
               .replace('','0')
               .fillna('0')
               .astype(float))'''
df['date'] = df['date'].replace(['Invalid','NaN',''],0).astype('datetime64[ns]')
```

```python
#Feature Eng  
bins = [0,100,500,float('inf')]
labels = ['Budget', 'Mid', 'Premium']
df['price_tier'] = pd.cut(df['price'],bins=bins, labels=labels)
df['days_since_order'] = (pd.Timestamp('today') - df['date']).dt.days
df['email_provider'] = df['domains'].str.split('.').str[0]
```

```python
df['price_tier'] = df.groupby('price_tier')['rating'].mean().round(2)
```
