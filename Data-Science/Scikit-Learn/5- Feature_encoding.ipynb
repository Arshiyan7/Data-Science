```python
from sklearn.datasets import fetch_openml
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
import pandas as pd 
```

# 1 Encoding Example 

```python
data = fetch_openml('car',as_frame=True).frame
```

```python
data
```

usually we can feed this type of categorical data to Models so there should be a way to convert these Categorical values in numerical because Models works well with numerical values, and the solution to this is to use feature encoding.

## what is feature Encoding? 
- Feature encoding is the process of converting categorical data (like text labels) into a numerical format so that machine learning models can understand and use it.
- e.g. assigns a number to each category (e.g., red=0, blue=1, green=2).

let's take some columns from our dataframe above and apply feature encoding

```python
columns_to_encode = ['lug_boot','safety']
encoder = OrdinalEncoder(
    categories=[
        ['small','med','big'],
        ['low','med','high']
    ]
)
data[columns_to_encode] = encoder.fit_transform(data[columns_to_encode])
```

```python
data
```

lug_boot and safety columns are now numerical and perfect to feed our model

```python
encoder.inverse_transform(data[columns_to_encode])
```

inverse_transform is used to turn numerical encoded columns back to categorical 

# 2 Encoding Example 

```python
data_2 = fetch_openml('adult',as_frame=True).frame
```

```python
data_2
```

```python
data_2.occupation.value_counts()
```

```python
encoder_2 = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
encoded_values = encoder_2.fit_transform(data_2[['occupation','race']])
new_cols = encoder_2.get_feature_names_out(['occupation','race'])
```

```python
encoded_values
```

```python
new_cols
```

```python
df_encoded = pd.DataFrame(encoded_values,index=data_2.index, columns=new_cols)

final_data = pd.concat(
    [data_2.drop(columns=['occupation','race']),
     df_encoded],
    axis=1
)
```

```python
final_data
```

occupation and race columns are gone and now we've seprate columns for each race and occupation with numerical values
