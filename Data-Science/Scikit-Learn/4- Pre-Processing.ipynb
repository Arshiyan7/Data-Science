## Pre-Processing: 
- preprocessing refers to a set of utilities and classes that help prepare raw data so it can be effectively used by machine learning models
- Many models assume that the input data has certain properties (e.g., all numeric, normalized, or free of missing values). Preprocessing transforms the data into a suitable format before feeding it into an estimator.
- Preprocessing in scikit-learn = preparing data (scaling, encoding, imputing, transforming) so that machine learning algorithms can learn from it effectively.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler # Mostly used 
```

```python
X, y = load_iris(return_X_y=True)
```

```python
X
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=11)
```

```python
scalar = StandardScaler()
X_train_Scaled = scalar.fit_transform(X_train)
X_test_Scaled = scalar.transform(X_test)
```
