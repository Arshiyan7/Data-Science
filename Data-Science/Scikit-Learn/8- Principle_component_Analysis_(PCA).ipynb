# PCA (Principle component Analysis):
an unsupervised dimensionality reduction technique that transforms high-dimensional data into a smaller set of new axes called principal components These components capture the maximum variance in the data, allowing us to project the data onto fewer dimensions while preserving as much information as possible. This helps simplify datasets, reduce noise, and improve the efficiency of machine learning models.

```python
from sklearn.datasets import fetch_openml
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
X, y = fetch_openml('mnist_784', return_X_y=True)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
pca = PCA(n_components=10)

X_test_reduced = pca.fit_transform(X_test)
X_train_reduced = pca.transform(X_train)
```

```python
X_train.shape
```

```python
X_train_reduced.shape
```

features of train reduced from 784 to 10 

```python
clf = LogisticRegression(max_iter=10)

clf.fit(X_train, y_train)
print(clf.score(X_test,y_test))
```

```python
clf = LogisticRegression(max_iter=10)

clf.fit(X_train_reduced, y_train)
print(clf.score(X_test_reduced,y_test))
```

performance decreased as components are decreased
