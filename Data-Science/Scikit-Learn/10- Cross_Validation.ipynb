# Cross-Validation 
Cross-validation is a technique to evaluate machine learning models more reliably by splitting the dataset into multiple folds, training on some folds, and testing on the remaining ones repeatedly. This provides a more accurate estimate of model performance compared to a single train-test split, which only tests the model once on one fixed portion of data. Unlike train-test split, cross-validation uses the data more efficiently and helps reduce overfitting, making it ideal for model selection and hyperparameter tuning. However, a final train-test split is often still used after cross-validation to get an unbiased evaluation on completely unseen data.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score
import numpy as np
```

```python
X, y = load_breast_cancer(return_X_y=True)

scalar = StandardScaler()
X_scaled = scalar.fit_transform(X)
```

```python
clf = KNeighborsClassifier()
scores = cross_val_score(clf, X_scaled, y, cv=5)
```

```python
print(scores)
print("Average performance: ",np.mean(scores))
```
