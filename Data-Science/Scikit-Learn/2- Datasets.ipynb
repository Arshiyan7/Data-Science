### 6 What is load, fetch, and make in sklearn.datasets?
- load_* → Small datasets that come with scikit-learn (e.g., breast cancer, iris)
- fetch_* → Larger datasets that are downloaded from the internet when needed
- make_* → Functions to generate synthetic datasets for practice/testing
- Example:
- from sklearn.datasets import load_iris, fetch_openml, make_classification
