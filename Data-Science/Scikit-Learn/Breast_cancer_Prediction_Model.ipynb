# Data preprocessing 

```python
X = cancer_df.drop(columns=['target'],axis=1)
y = cancer_df['target']
```

## Train and test Split - Standarization

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=11, test_size=0.2)
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
X_train_sc = sc.fit_transform(X_train)
X_test_sc = sc.transform(X_test)
```

### Training = read (fit) + apply (transform)
- Here it’s not the model that reads the rules, but the scaler (preprocessor).
- The scaler learns the “rules” (mean, std) from the training data → then applies them to training features.

### Testing = just apply (transform), no re-reading (fit)
- The scaler applies the exact same rules learned from training to the test data.
- This ensures both train and test are on the same scale, but without leaking test info.

## Models Training 

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
```

```python
### SVC Model (without scaled data)
from sklearn.svm import SVC
sv = SVC()
sv.fit(X_train, y_train)
y_pred_svc = sv.predict(X_test)
accuracy_score(y_pred_svc,y_test)
```

```python
### SVC Model (with scaled data)
from sklearn.svm import SVC
sv_2 = SVC()
sv_2.fit(X_train_sc, y_train)
y_pred_2_svc = sv_2.predict(X_test_sc)
accuracy_score(y_pred_2_svc,y_test)
```

```python
### Log regression (train without sclaed data)
from sklearn.linear_model import LogisticRegression
log = LogisticRegression()
log.fit(X_train, y_train)
y_pred_log = log.predict(X_test)
accuracy_score(y_pred_log,y_test)
```

```python
### Log regression (train with sclaed data)
from sklearn.linear_model import LogisticRegression
log = LogisticRegression()
log.fit(X_train_sc, y_train)
y_pred_log_2 = log.predict(X_test_sc)
accuracy_score(y_pred_log_2,y_test)
```

```python
### K-Neighbors (train without sclaed data)
from sklearn.neighbors import KNeighborsClassifier
KN = KNeighborsClassifier()
KN.fit(X_train, y_train)
y_pred_KN = KN.predict(X_test)
accuracy_score(y_pred_KN,y_test)
```

```python
### K-Neighbors (train with sclaed data)
from sklearn.neighbors import KNeighborsClassifier
KN = KNeighborsClassifier()
KN.fit(X_train_sc, y_train)
y_pred_KN_2 = KN.predict(X_test_sc)
accuracy_score(y_pred_KN_2,y_test)
```

```python
### RandomForestClassifier (train without sclaed data)
from sklearn.ensemble import RandomForestClassifier
Random_fr = RandomForestClassifier()
Random_fr.fit(X_train, y_train)
y_pred_rf = Random_fr.predict(X_test)
accuracy_score(y_pred_rf,y_test)
```

```python
### RandomForestClassifier (train with sclaed data)
from sklearn.ensemble import RandomForestClassifier
Random_fr = RandomForestClassifier()
Random_fr.fit(X_train_sc, y_train)
y_pred_rf_2 = Random_fr.predict(X_test_sc)
accuracy_score(y_pred_rf_2,y_test)
```

```python
### DecisionTreeClassifier (train without sclaed data)
from sklearn.tree import DecisionTreeClassifier
D_tree = DecisionTreeClassifier()
D_tree.fit(X_train, y_train)
y_pred_D_tree = D_tree.predict(X_test)
accuracy_score(y_pred_D_tree,y_test)
```

```python
### DecisionTreeClassifier (train with sclaed data)
from sklearn.tree import DecisionTreeClassifier
D_tree = DecisionTreeClassifier()
D_tree.fit(X_train_sc, y_train)
y_pred_D_tree_2 = D_tree.predict(X_test_sc)
accuracy_score(y_pred_D_tree_2,y_test)
```

### notice how scaled data gives off better predicition than non-scaled data

### After comparing standardized and non-standardized data, it’s clear that scaling plays a major role in models that rely on distances or gradients such as:
- KNN, Logistic Regression, and SVM. These models showed higher accuracy after standardization because equalizing feature scales helps them learn more balanced and effective decision boundaries.

- In contrast, tree-based models like Decision Tree and Random Forest produced nearly identical results with or without standardization. This is because they split data based on feature thresholds, not distance or magnitude, making them naturally insensitive to feature scale.

**Overall, standardization is essential for most mathematical and distance-based algorithms but has little to no impact on tree-based models.**


## Model Confusion Matrix

<img src='https://media.geeksforgeeks.org/wp-content/uploads/20250530174111874391/predicted_condition_2_.webp' />

### in the above image:  
- **True Positive (TP)**: The model correctly predicted a positive outcome i.e the actual outcome was positive.
- **True Negative (TN)**: The model correctly predicted a negative outcome i.e the actual outcome was negative.
- **False Positive (FP)**: The model incorrectly predicted a positive outcome i.e the actual outcome was negative. It is also known as a Type I error.
- **False Negative (FN)**: The model incorrectly predicted a negative outcome i.e the actual outcome was positive. It is also known as a Type II error.

If the diagonal values are high and off-diagonal ones are low, the model is performing well. In short more correct predictions = stronger diagonal = better accuracy.

## Hyperparameter Tuning using RandomizedSearchCV

```python
param_dist = {
    'n_estimators': [100, 200, 300, 400, 500],        
    'max_depth': [4, 6, 8, 10, 12, None],             
    'min_samples_split': [2, 5, 10],                  
    'min_samples_leaf': [1, 2, 4],                   
    'max_features': ['sqrt', 'log2'],                 
    'bootstrap': [True, False]                        
}
```

```python
from sklearn.model_selection import RandomizedSearchCV
rf = RandomForestClassifier(random_state=42)

rf_random = RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_dist,
    n_iter=20,                
    cv=5,                     
    verbose=2,
    random_state=42,
    n_jobs=-1,               
    scoring='accuracy'
)
```

```python
rf_random.fit(X_train_sc, y_train)
```

```python
print("Best Parameters:", rf_random.best_params_)
print("Best Cross-Validation Accuracy:", rf_random.best_score_)
```

```python
best_rf = rf_random.best_estimator_
y_pred_best = best_rf.predict(X_test_sc)
```

### Evaluation of Tuned Model (Confusion Matrix + Classification Report)

```python
print(classification_report(y_test,y_pred_rf_2)) #Before Tuning 
```

```python
print(confusion_matrix(y_test, y_pred_best))
print(classification_report(y_test, y_pred_best)) # After Tuning 
```

### Before and After Tuning Comparison

Before tuning, the **Random Forest model** performed exceptionally well with an accuracy of **99%**, showing excellent precision and recall across both classes. It handled the dataset effectively, maintained balance between classes, and produced minimal misclassifications.

After applying **RandomizedSearchCV** for hyperparameter tuning, the model’s accuracy slightly decreased to **97%**. Although this appears as a drop in performance, it indicates that the tuned model became **more regularized and stable**, reducing the chance of overfitting to a specific training-test split.

In summary:

* **Before tuning:** Higher accuracy, slightly more overfit to the given data.
* **After tuning:** Slightly lower accuracy, but improved generalization and reliability for unseen data.

Thus, while the tuned Random Forest didn’t outperform the base version numerically, it is likely to perform **more consistently** across different datasets — a valuable trait in real-world machine learning scenarios.
