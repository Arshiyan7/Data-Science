```python
#1 -> we are importing breast cancer dataset using sklearn datasets
#2 -> allows to splits data into training and test portion
#3 -> Scales data it is useful cause some data/datasets are scale sensitive so we use this for those datasets 
#4 -> Used to learn from Data and make predictions 
```

```python
#1 
from sklearn.datasets import load_breast_cancer 
#2
from sklearn.model_selection import train_test_split 
#3
from sklearn.preprocessing import StandardScaler
#4
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Create a dataset  
data = load_breast_cancer(as_frame=True).frame
```

```python
# load the dataset
data
```

```python
#1 Divide the dataset in X and y (More preferable format for Sklearn modeling)
X,y = load_breast_cancer(return_X_y=True)
```

```python
X
```

here X is returned as NumPy array and X shows all the features in the dataset 

```python
y
```

y shows all of the target values in the dataset 

X and y datasets are the format that we need for scikit learn modeling and it is generally recommended! 

```python
#2 take X and y dataset and split it to training and testing data 
X_train, X_test, y_train, y_test = train_test_split(X,y, train_size=0.2)
```

```python
#3 Scale the data by passing X_train and X_test 
scaler = StandardScaler()

X_train_Scaled = scaler.fit_transform(X_train)
X_test_Scaled = scaler.transform(X_test)
```

```python
#4 Train the model 
knn = KNeighborsClassifier()

knn.fit(X_train_Scaled, y_train)
```

```python
print(knn.score(X_test_Scaled, y_test))
```
