# Metrices
In scikit-learn, metrics are used to evaluate how well a machine learning model performs by comparing predictions to actual values. For classification, it provides measures like accuracy, precision, recall, F1-score, ROC-AUC, and confusion matrices. For regression, it offers MSE, MAE, and R² score to assess prediction errors. These metrics help choose and tune models effectively based on the problem’s needs.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
```

```python
X, y = load_breast_cancer(return_X_y=True)

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2)

scalar = StandardScaler()
X_train_scaled = scalar.fit_transform(X_train)
X_test_scaled = scalar.fit_transform(X_test)
```

```python
clf = KNeighborsClassifier()
clf.fit(X_train_scaled,y_train)
```

```python
clf.score(X_test_scaled,y_test) #Accuracy 
```

```python
#Classification task
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score

y_pred = clf.predict(X_test_scaled)
```

```python
accuracy_score(y_pred,y_test)
```

```python
f1_score(y_pred,y_test)
```

```python
recall_score(y_pred,y_test)
```

```python
precision_score(y_pred,y_test)
```

```python
# Regression task
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
reg = LinearRegression()
reg.fit(X_train_scaled,y_train)
```

```python
y_pred = reg.predict(X_test_scaled)
```

```python
mean_absolute_error(y_pred,y_test)
```

```python
r2_score(y_pred,y_test)
```

```python
mean_squared_error(y_pred,y_test)
```
